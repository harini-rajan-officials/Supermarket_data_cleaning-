```python
import pandas as pd
```

```python
df = pd.read_csv("supermarket_data.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df['Gender'] = df['Gender'].str.strip().str.lower()
df['Gender'] = df['Gender'].replace({'m': 'male', 'f': 'female'})
```

```python
df['Gender'] = df['Gender'].str.strip().str.lower()
df['Gender'] = df['Gender'].replace({'m': 'male', 'f': 'female'})
```

```python
df['Date'].head()
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y')

```

```python
df['Date'] = df['Date'].dt.strftime('%d-%m-%Y')
```

```python
df.head()
```

```python
df.columns = df.columns.str.strip()        # Remove spaces
df.columns = df.columns.str.lower()        # Lowercase
df.columns = df.columns.str.replace(' ', '_')  # Replace spaces with underscore
```

```python
df.head()
```

```python
df.dtypes
```

```python
df['date'] = pd.to_datetime(df['date'], format='%d-%m-%Y')
```

```python
df['quantity'] = df['quantity'].astype(int)
df['unit_price'] = df['unit_price'].astype(float)
df['time'] = pd.to_datetime(df['time'])

```

```python
df.dtypes
```

```python
df['gender'] = df['gender'].str.strip()   # remove extra spaces
df['gender'] = df['gender'].str.lower()   # convert to lowercase
```

```python
df['branch'] = df['branch'].str.strip().str.lower()
df['city'] = df['city'].str.strip().str.lower()
df['product_line'] = df['product_line'].str.strip().str.lower()
df['customer_type'] = df['customer_type'].str.strip().str.lower()
df['payment'] = df['payment'].str.strip().str.lower()
```

```python
df.head()
```

```python
df.to_csv("cleaned_supermarket_data.csv", index=False)
```
